```python
import matplotlib.pyplot as plt
import numpy as np
import random
```

```python
x = np.array([2023,2024, 2025, 2026])
y = np.array([15, 25, 30, 20])

plt.plot(x,y)
plt.show()
```

```python
x = np.array([2023,2024, 2025, 2026])
y = np.array([15, 25, 30, 20])

plt.plot(x,y,marker="o", markersize = 10, markerfacecolor = "Red",markeredgecolor = "Black", linestyle = 'dotted', linewidth = 4, color = "#1707f2")
plt.show()

# linestyle is by-default  solid
# markersize can be written like 'ms'
# markerfacecolor can be writte lik "mfc"
# markeredgecolor can be writte lik "mec"
```

```python
x = np.array([2023,2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38,5])

plt.plot(x,y1,marker="o", markersize = 10, markerfacecolor = "Red",markeredgecolor = "Black", linestyle = 'dotted', linewidth = 4, color = "#1707f2")
plt.plot(x,y2,marker="o", markersize = 10, markerfacecolor = "Red",markeredgecolor = "Black", linestyle = 'solid', linewidth = 4, color = "#1407f2")
plt.show()
```

```python
y3 = np.array([1,2,3,4])
line_style = dict(
    marker="o",
    markersize = 10,
    markerfacecolor = "Red",
    markeredgecolor = "Black",
    linestyle = 'solid', #ls
    linewidth = 4, # lw

)
plt.plot(x,y1,color = "#f507a6", **line_style)
plt.plot(x,y2,color = "#1707f2", **line_style)
plt.plot(x,y3,color = "#24d609", **line_style)

plt.show()
```

```python
x = np.array([2023,2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38,5])
y3 = np.array([13, 15, 20, 30])

plt.title("Class Size",
          fontsize = 25,
          fontweight = 'bold',
          color = "#6905ff")


plt.xlabel("Year", fontsize = 20)
plt.ylabel("Students", fontsize = 20)

plt.plot(x, y1)
plt.plot(x, y2)
plt.plot(x, y3)

plt.xticks(x)
plt.show()
```

```python
x = [1,2,3,4,5]
y = [5,10,15,20,25]

plt.grid(axis = "x",
         linewidth = 2,
         color = "lightgray",
         linestyle = "dashed")
plt.grid(axis = "y",
         linewidth = 2,
         color = "lightgray",
         linestyle = "dashed")
plt.plot(x,y)
plt.show()
```

```python
categories = np.array(["grains","fruit","vegetables","protein","dairy","sweets"])
values = np.array([4,3,2,5,3,1])

plt.bar(categories,values, color = "#c9a6ff")

plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```

```python
categories = np.array(["grains","fruit","vegetables","protein","dairy","sweets"])
values = np.array([4,3,2,5,3,1])

plt.barh(categories,values, color = "#3d0e82")

plt.title("Daily Consumption")
plt.xlabel("Food")
plt.ylabel("Quantity")
plt.show()
```

```python
categories = np.array(["Freshmen", "Sophmores", "Juniors", "Seniors"])
values = np.array([300, 250, 275, 225])
colors = ['#decafc', '#400f8a', '#870ccf', '#ea94f2']

plt.pie(values, labels = categories,
        autopct = "%1.1f%%",
        colors = colors)

plt.title("Garden of Eden")
plt.show()
```

```python
categories = np.array(["Freshmen", "Sophmores", "Juniors", "Seniors"])
values = np.array([300, 250, 275, 225])
colors = ['#decafc', '#400f8a', '#870ccf', '#ea94f2']

plt.pie(values, labels = categories,
        autopct = "%1.1f%%",
        colors = colors,
        explode = [0, 0 , 0, 0.1],
        shadow = True,
        startangle = 90)

plt.title("Garden of Eden")
plt.show()
```

```python
x = [0,1,1, 2,3,4,5,6,7,7,8]
y = [55, 60, 65,62, 68, 70, 75,78,82,85, 87]

plt.scatter(x, y, color = "skyblue",
            alpha = 0.5,  # tarnsparency
            s = 100 #size
            )
plt.show()
```

```python
x1 = [0,1,1, 2,3,4,5,6,7,7,8]
y1 = [55, 60, 65,62, 68, 70, 75,78,82,85, 87]

x2 = [0,1,2,2,3,4,5,6,7,8,9]
y2 = [50, 58, 65,70, 72, 80, 75,68,82,95, 87]

plt.scatter(x1, y1, color = "skyblue",
            alpha = 0.5,  # tarnsparency
            s = 100, #size
            label = "Class A")

plt.scatter(x2, y2, color = "red",
            alpha = 0.5,  # tarnsparency
            s = 100, #size
            label = "Class B")

plt.title("Test Scores")
plt.xlabel("Hours Studied")
plt.ylabel("Grades")

plt.legend()
plt.show()
```

```python
scores = np.random.normal(loc= 80, scale = 10, size = 100)

scores = np.clip(scores, 0, 100)
plt.hist(scores, bins =10, color = "#400f8a",
         edgecolor = "pink")

plt.title("Exam Scores")
plt.xlabel("Score")
plt.ylabel("# of students")

plt.show()
```

```python
figure, axes = plt.subplots(2,2)

x = np.array([1,2,3,4,5])
axes[0,0].plot(x, x*2, color = "red")
axes[0,0].set_title("x*2")
axes[0,1].plot(x, x**2, color = "blue")
axes[0,1].set_title("x**2")
axes[1,0].bar(x, x**3, color = "green")
axes[1,0].set_title("x**3")
axes[1,1].plot(x, x**4, color = "purple")
axes[1,1].set_title("x**4")
plt.tight_layout()
```
